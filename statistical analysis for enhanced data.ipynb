```python
def mode(data):

    counts = {}
    for x in data:
        counts[x] = counts.get(x, 0) + 1

    max_count = max(counts.values())
    mode = [k for k, v in counts.items() if v == max_count]

    if len(mode) == 1:
        return mode[0]
    else:
        return mode
```

```python
mode_original = mode (data)
```

```python
mode_original
```

```python
def quartiles(data):

    mid = len(data) // 2

    first_half = data[:mid]
    second_half = data[mid:] if len(data) % 2 == 0 else data[mid+1:]

    first_mid = len(first_half) // 2
    if len(first_half) % 2 == 0:
        Q1 = (first_half[first_mid - 1] + first_half[first_mid]) / 2
    else:
        Q1 = first_half[first_mid]

    second_mid = len(second_half) // 2
    if len(second_half) % 2 == 0:
        Q3 = (second_half[second_mid - 1] + second_half[second_mid]) / 2
    else:
        Q3 = second_half[second_mid]

    return Q1, Q3
Q2 = median 
```

```python
Q1, Q3 = quartiles(data)
```

```python
print (f'Q1 is {Q1} and Q3 is {Q3}')
```

```python
IQR = Q3 - Q1 
IQR
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
lower_extreme = Q1 - 3 * IQR
upper_extreme = Q3 + 3 * IQR
```

```python
moderate = []
high = []
extreme = []
```

```python
for x in data:
    score = 0
    if x > upper_bound:
        score = (x - Q3) / IQR
    elif x < lower_bound:
        score = (Q1 - x) / IQR
    else:
        continue

    if 1.5 <= score < 3:
        moderate.append(x)
    elif 3 <= score <= 5:
        high.append(x)
    elif score > 5:
        extreme.append(x)
```
